# "Introduction to Random Strings" solution optimization using NumPy

## Data set:

````python
dna_seq = "GAATTATGGAAATAGTCATCAAGCTTGAGTCTGGACTCAACACTTAGGAAAATCCTATAGGAGCTGGGGTTCAAACATATCGCAA"
cg_content = "0.084 0.136 0.164 0.214 0.291 0.339 0.354 0.442 0.460 0.533 0.567 0.636 0.699 0.721 0.758 0.807 0.853 0.929"
huge_cg_content = " ".join([cg_content] * 1000)
````

## Solution using vanilla python:

````python
from math import log10

def vanilla_solution(dna_seq, cg_content_string):
    cg_content_list = [float(item) for item in cg_content_string.split(' ')]
    result = []

    cg_count = dna_seq.count('G') + dna_seq.count('C')
    at_count = len(dna_seq) - cg_count
    
    for cg_prob in cg_content_list:
        prob = (cg_count * log10(cg_prob / 2)) + (at_count * log10((1 - cg_prob) / 2))
        result.append(prob)
    return result
````

- Time complexity: $O(N)$
- Space complexity: $O(N)$

````python
%timeit vanilla_solution(dna_seq, huge_cg_content)
````

````python
%timeit vanilla_solution(dna_seq, cg_content)
````

````python
answer = vanilla_solution(dna_seq, cg_content)
print(answer)
````

## Solution using numpy:

````python
import numpy as np
from math import log10

def numpy_solution(dna_seq, cg_content_string):
    gen = (float(item) for item in cg_content_string.split(' '))
    cg_content_arr = np.fromiter(gen, dtype=float)

    gc_count = dna_seq.count('G') + dna_seq.count('C')
    at_count = len(dna_seq) - gc_count

    cg_probs = np.log10(cg_content_arr / 2)
    at_probs = np.log10((1 - cg_content_arr) / 2)
    
    result = (gc_count * cg_probs) + (at_count * at_probs)

    return result.tolist()
````

- Time complexity: $O(N)$
- Space complexity: $O(N)$

````python
%timeit numpy_solution(dna_seq, huge_cg_content)
````

````python
%timeit numpy_solution(dna_seq, cg_content)
````

````python
answer = numpy_solution(dna_seq, cg_content)
print(answer)
````

**Summary:** Vanilla Python outperforms NumPy when processing small inputs (e.g., the original ```cg_content``` string). In this case, Python took approximately **26.8 μs** compared to NumPy's **52.7 μs**. This is because the overhead of initializing NumPy arrays, converting data types, and calling C-extensions outweighs the actual computation time for a few elements. NumPy demonstrates its true power on the "huge" dataset (scaled by 1000x). The NumPy implementation completed the task in **14.6 ms**, while the vanilla solution took **33.2 ms** making NumPy more than twice as fast.
