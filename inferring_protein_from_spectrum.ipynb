## Constants:

```python
import numpy as np

AMINO_ACIDS_MASS = {
        "A": 71.03711,
        "C": 103.00919,
        "D": 115.02694,
        "E": 129.04259,
        "F": 147.06841,
        "G": 57.02146,
        "H": 137.05891,
        "K": 128.09496,
        "L": 113.08406,
        "M": 131.04049,
        "N": 114.04293,
        "P": 97.05276,
        "Q": 128.05858,
        "R": 156.10111,
        "S": 87.03203,
        "T": 101.04768,
        "V": 99.06841,
        "W": 186.07931,
        "Y": 163.06333
    }

AMINO_LABELS = np.array(list(AMINO_ACIDS_MASS.keys()))
AMINO_VALUES = np.array(list(AMINO_ACIDS_MASS.values()), dtype=np.float64)
```

## Solution using vanilla python:

```python
def vanilla_solution(mass_string, mass_table):        
    mass_list = [float(mass.strip()) for mass in mass_string.strip().splitlines()]
    protein_string = ""
    
    for i in range(len(mass_list) - 1):
        protein_mass = abs(mass_list[i] - mass_list[i + 1])
        closest_mass_protein = ''
        for protein, mass in mass_table.items():
            if abs(protein_mass - mass) < 0.0001:
                closest_mass_protein = protein
        protein_string += closest_mass_protein
    return protein_string
```

- Time complexity: $O(N)$
- Space complexity: $O(N)$

```python
%timeit vanilla_solution(data_set, AMINO_ACIDS_MASS)
```

```python
answer = vanilla_solution(data_set, AMINO_ACIDS_MASS)
print(answer)
```

## Solution using numpy:

```python
import numpy as np

def numpy_solution(mass_string, labels_arr, values_arr):
    gen = (float(mass.strip()) for mass in mass_string.strip().splitlines())
    mass_arr = np.fromiter(gen, dtype=float)
    diffs = np.abs(np.diff(mass_arr))
    indices = np.abs(diffs[:, None] - values_arr).argmin(axis=1)
    protein_string = "".join(labels_arr[indices])
    return protein_string
```

- Time complexity: $O(N)$
- Space complexity: $O(N)$

```python
%timeit numpy_solution(data_set, AMINO_LABELS, AMINO_VALUES)
```

```python
answer = numpy_solution(data_set, AMINO_LABELS, AMINO_VALUES)
print(answer)
```

**Summary:** NumPy significantly outperforms vanilla Python even on this moderately-sized dataset. The NumPy implementation completed in **179 μs**, while the vanilla solution took **350 μs** — making NumPy more than ~2x faster. This is because the vectorized broadcasting operation (`diffs[:, None] - values_arr`) eliminates the nested loop overhead inherent in the vanilla approach.
